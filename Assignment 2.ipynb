```python
from serpapi import EbaySearch

all_results = []

# Search 1 - hockey cards (sold listings)
params = {
    "engine": "ebay",
    "_nkw": "hockey cards",
    "ebay_domain": "ebay.com",
    "api_key": "12a13973b8d58a4cc7b569859b257ef029573e1b3f116a9ecfcebd95b1f016a5",
    "show_only": "Sold",
    "lh_sold": "1"
}

search = EbaySearch(params)
results = search.get_dict()
if "organic_results" in results:
    all_results.extend(results["organic_results"])

# Search 2 - hockey card hobby boxes (sold listings)
params["_nkw"] = "hockey card hobby boxes"
search = EbaySearch(params)
results = search.get_dict()
if "organic_results" in results:
    all_results.extend(results["organic_results"])

# filter for sold items over $100
expensive = [
    item for item in all_results
    if isinstance(item.get("price"), dict)
    and item["price"].get("extracted", 0) > 100
    and item.get("sold_date") is not None
]

print(f"Total items: {len(all_results)}")
print(f"Sold items over $100: {len(expensive)}")

for item in expensive:
    print(f"Title: {item['title']}")
    print(f"Price: {item['price']['raw']}")
    print(f"Sold: {item['sold_date']}")
    print("---")
```
